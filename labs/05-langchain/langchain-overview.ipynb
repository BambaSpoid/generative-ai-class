#  LangChain — Ecosystem, Architecture & Enterprise Ready

---

## Pourquoi LangChain ?

Avant LangChain :

* OpenAI → SDK spécifique
* HuggingFace → Transformers
* Gemini → autre SDK
* Llama → autre interface

Chaque modèle = API différente
Chaque projet = code différent

`Résultat` :

* Code non standardisé
* Fort couplage au provider
* Difficile à maintenir

---

### La solution : LangChain

LangChain est :

>  Un framework open-source pour développer des applications GenAI.

Il apporte :

* Abstraction des LLMs
* Composition modulaire
* Intégration RAG
* Support Agents
* Chemin vers la production

---

## LangChain vs LlamaIndex

#### Différence intuitive

* LlamaIndex → centré sur les données et le RAG
* LangChain → centré sur l’orchestration d’applications GenAI complètes

---

#### Comparaison

| Critère   | LangChain              | LlamaIndex                     |
| --------- | ---------------------- | ------------------------------ |
| Focus     | Orchestration complète | RAG & Indexation               |
| Agents    | Très fort              | Possible mais secondaire       |
| RAG       | Très bon               | Excellent                      |
| LLMOps    | LangSmith intégré      | Intégrations externes          |
| API       | LangServe              | FastAPI custom                 |
| Cas idéal | App GenAI complexe     | Recherche documentaire avancée |

---

#### Règle pratique

* RAG pur documentaire → LlamaIndex
* Application complète (RAG + agents + tools + API) → LangChain
* En entreprise → souvent les deux combinés

###  MODULE 1 — Ingestion & Indexation des données

👉 Objectif : Préparer les données pour qu’elles soient interrogeables efficacement.

---

#### 1.  Data Ingestion (Chargement des données)

`Premier composant clé de LangChain`.

Permet de charger des données depuis différentes sources :

* PDF
* Excel
* JSON
* Images
* Vidéos
* URLs
* Bases de données

➡️ C’est la porte d’entrée du pipeline.

---

#### 2. Text Splitting (Découpage en chunks)

Pourquoi ?

Les LLMs ont une **limite de contexte** (context window).

Donc on :

* Divise les documents volumineux
* Crée des chunks plus petits
* Prépare le texte pour traitement

`Composant clé` : **Text Splitters**

C’est une étape de **data transformation**.

---

#### 3. Embeddings (Vectorisation)

On convertit les chunks texte en **vecteurs numériques**.

Pourquoi ?

Parce que la recherche s’effectue via :

* Similarity Search
* Cosine Similarity

`Différents types d’embeddings` :

* OpenAI embeddings
* Gemini embeddings
* HuggingFace embeddings
* Modèles open source

---

#### 4. Vector Store (Base de données vectorielle)

Les vecteurs doivent être stockés.

`Exemples de Vector DB` :

* FAISS
* ChromaDB
* AstraDB
* Pinecone (généralement)

Rôle :

* Stocker les vecteurs
* Permettre la recherche par similarité
* Retourner le **contexte pertinent**

---

### MODULE 2 — Interrogation & Génération

👉 Objectif : Répondre aux questions utilisateurs.

---

#### 5. User Query

L’utilisateur pose une question.

---

#### 6. Prompt Template

On définit :

* Le comportement du modèle
* Son rôle
* Son style

Exemple :

> “Agis comme un chercheur en IA…”

`Composant clé` : **Prompt Templates**

---

#### 7. Retrieval Chain

Composant très important.

Responsable de :

* Interroger le Vector Store
* Récupérer le contexte pertinent
* Injecter ce contexte dans le prompt

`Concepts associés` :

* Document Store
* Retriever
* Retrieval Chain

---

#### 8. LLM (Génération finale)

Entrées :

* Prompt Template
* Question utilisateur
* Contexte récupéré

Sortie :

* Réponse ancrée sur les documents

---
