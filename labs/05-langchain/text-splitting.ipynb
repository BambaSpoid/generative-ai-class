
# Étape 2: Text Splitting 

- Objectif : découper les documents en *chunks* (morceaux) exploitables par un LLM.

Nous utilisons **RecursiveCharacterTextSplitter**, recommandé pour du texte générique.
Il essaie de découper en respectant l’ordre suivant :
1) Paragraphes ("\n\n")
2) Lignes ("\n")
3) Espaces (" ")
4) Caractères ("") si nécessaire

`Paramètres clés` :
- `chunk_size` : taille max d’un chunk (en caractères)
- `chunk_overlap` : chevauchement entre chunks (pour garder le contexte)


#### Requirements (Text Splitting)

### 1. Split sur des Documents PDF

---

- `Input` : une liste de `Document` (ex: sortie de PyPDFLoader)
- `Output` : une liste de `Document` découpés en chunks

On va :
1) Charger `attention.pdf`
2) Appliquer `RecursiveCharacterTextSplitter`
3) Inspecter les chunks


```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = PyPDFLoader("attention.pdf")
docs = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

final_documents = text_splitter.split_documents(docs)

print("Nombre total de chunks :", len(final_documents))
print("\n--- Chunk 1 ---\n", final_documents[0].page_content[:400])
print("\nMetadata chunk 1 :", final_documents[0].metadata)
```

#### Inspection (chunks 2)

On affiche 2 chunks pour voir :
- le découpage réel
- le chevauchement (overlap)


```python
print("----- CHUNK 0 -----")
print(final_documents[0].page_content)

print("\n----- CHUNK 1 -----")
print(final_documents[1].page_content)
```

## Take Away (Text Splitting)

- Le LLM ne peut pas lire un énorme document d’un coup → limitation de contexte.
- On découpe en chunks + overlap pour préserver le contexte.
- `split_documents()` : découpe une liste de `Document`
- `create_documents()` : crée des `Document` à partir de texte brut
