
#  LoRA

Si vous prenez un de ces LLMs:

* OpenAI (GPT-4)
* Meta (LLaMA)
* Google (Gemma)

Il contient **des milliards de poids**.

Si tu veux l’adapter à :

* Finance
* Médecine
* Chatbot entreprise
* Text-to-SQL

👉 La méthode classique dit :

> On modifie TOUS les poids.

Problème ?

* GPU énorme requis
* VRAM énorme
* Coût cloud énorme
* Stockage énorme

---

## L’idée géniale de LoRA

LoRA dit :

-   > ❌ Ne modifie pas tous les poids
-   > ✅ Ajoute une petite “couche d’adaptation”

Au lieu de modifier la grande matrice du modèle…

On ajoute **deux petites matrices**.

---

## 🎨 MindMap Simple

Avant :

```
W (énorme matrice)
```

Fine-tuning classique :

```
On change W entièrement
```

Avec LoRA :

```
W (reste gelé)
+
Petite adaptation = A × B
```

Donc le modèle final devient :

```
W + (A × B)
```

---

## Pourquoi c’est intelligent ?

Parce que :

* W = milliards de paramètres
* A et B = quelques millions seulement

Donc :

* Beaucoup moins de mémoire
* Beaucoup moins de GPU
* Beaucoup plus rapide

---

## Le secret : “Low Rank”

LoRA signifie :

> Low Rank Adaptation

Ça veut dire :

Au lieu de stocker une grande matrice complète,
on la représente comme la multiplication de deux petites matrices.

👉 C’est comme reconstruire un grand Lego avec deux petits blocs.

---

## Intuition encore plus simple

Imagine que ton modèle sait déjà parler anglais.

Tu veux juste :

* Le spécialiser en finance.

Tu n’as pas besoin de lui réapprendre l’anglais.

Tu ajoutes juste une petite correction ciblée.

C’est exactement ce que fait LoRA.

---

##  Donc LoRA, c'est :

* On garde le modèle gelé
* On ajoute une petite adaptation
* On entraîne seulement cette petite partie
* On économise énormément de ressources

---

#  Full Fine-Tuning (le problème)

--- 

Dans le fine-tuning classique :

* On met tous les paramètres en “trainable”
* On met à jour tous les gradients
* On stocke tous les gradients
* On sauvegarde tous les nouveaux poids

Pour un modèle 7B ou 70B :
→ Très coûteux.

---

## Ce que fait LoRA exactement

On garde les poids pré-entraînés :

[
W_0
]

On ajoute une mise à jour :

[
\Delta W = B \times A
]

Donc :

[
W = W_0 + BA
]

Mais :

* (W_0) est gelé
* Seuls A et B sont entraînés

---

## Pourquoi ça réduit les paramètres

Une matrice complète :

* taille = d × k

LoRA la remplace par :

* A : d × r
* B : r × k

où r = rank (petit nombre comme 4, 8, 16)

Donc :

d × k → devient → r(d + k)

Si r est petit :
→ énorme réduction de paramètres.

---

##  Paramètre clé : le Rank

* Rank faible → moins de paramètres → plus rapide
* Rank élevé → plus expressif → plus précis

En pratique :

* r = 4 ou 8 fonctionne très bien
* r élevé pour tâches complexes

---

##  Avantages concrets

Avec LoRA :

-    100× moins de paramètres entraînés
-    VRAM divisée
-   Possibilité fine-tuning sur 1 GPU
-   Stockage léger (on stocke juste A et B)

---

## Impact industriel

LoRA permet :

* Domain fine-tuning
* Task-specific adaptation
* Rapid experimentation
* Multiple adapters sur un même modèle

C’est devenu standard en industrie.

#  QLoRA

---

Maintenant imagine :

LoRA était déjà intelligent.

QLoRA dit :

> Et si on combinait LoRA + Quantization ?

---

## Rappel rapide

Quantization :

* 32 bits → 8 bits → 4 bits
* Réduit mémoire

LoRA :

* Ne met à jour que petites matrices

---

## 💡 QLoRA combine les deux

QLoRA fait :

1. On quantize le modèle en 4-bit
2. On garde ce modèle gelé
3. On applique LoRA dessus

Résultat :

-   Fine-tuning possible sur GPU normal
-   Très faible mémoire
-   Très efficace

---

##  MindMap Simple

Avant :

```
Modèle énorme en 32-bit
```

QLoRA :

```
Modèle compressé en 4-bit
+
Petite adaptation LoRA
```

Ultra léger.

---

## Le problème même avec LoRA

Même si LoRA réduit les paramètres entraînés :

Le modèle de base est encore en FP16.

Donc il occupe encore beaucoup de mémoire.

---

##  Ce que fait QLoRA

QLoRA :

* Charge le modèle en 4-bit (quantized)
* Garde les poids quantized gelés
* Entraîne seulement les matrices LoRA en FP16

Donc :

* Base model = 4-bit
* LoRA adapters = 16-bit

---

## Pourquoi c’est puissant

Mémoire :

* Modèle 7B en FP16 ≈ 14GB
* Modèle 7B en 4-bit ≈ ~3.5GB

Donc :

Fine-tuning 7B possible sur 1 GPU 16GB.

C’est une révolution.

---

## Point important

QLoRA utilise :

* Quantization 4-bit
* Méthode spécifique pour éviter perte trop forte
* Reconstruction interne en 16-bit pendant calcul

Donc :

* Compression en stockage
* Stabilité en calcul

---

##  Résumé final LoRA vs QLoRA

| Technique | Ce qui change             |
| --------- | ------------------------- |
| Full FT   | Tout est mis à jour       |
| LoRA      | Petites matrices ajoutées |
| QLoRA     | Modèle quantized + LoRA   |
